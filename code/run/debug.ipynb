```python
import os
def cal_kstar(input_dir:str,output_same_kstar:str, output_same_costheta:str, output_mix_kstar_dir:str, mixing_time: int=21):
    input_path = os.path.join(input_dir, "*.parquet")

    X = ak.from_parquet(input_path)
    same_pair = gen_pair(X)
    kstar_same = kstar(same_pair)
    cosAngle = cos_theta(same_pair)
    
    ak.to_parquet(kstar_same, output_same_kstar)
    ak.to_parquet(cosAngle, output_same_costheta)
    for i in range(1, 21):
        mixing_pair = gen_mixed_pair(X)
        kstar_mix = kstar(mixing_pair)
        ak.to_parquet(kstar_mix, f"{output_mix_kstar_dir}/kstar_mix{i}.parquet")
```
